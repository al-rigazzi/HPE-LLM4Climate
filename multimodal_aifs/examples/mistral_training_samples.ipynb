# Mistral Training Samples - Full Prompts and Responses

This notebook demonstrates the training pipeline by generating climate data samples and showing the complete prompts (including tabulated statistics) and Mistral's responses.

The pipeline:
1. Loads climate data from Zarr files
2. Selects random geographic locations
3. Computes statistics (min/max/mean/gradient/rotation/divergence)
4. Formats statistics as tables
5. Generates prompts for Mistral
6. Gets real Mistral-7B-Instruct-v0.3 responses

```python
# Import required libraries
import sys
from pathlib import Path

# Add parent directory to path
sys.path.insert(0, str(Path.cwd().parent.parent))

from multimodal_aifs.training import ClimateTextDataLoader
```

## Initialize the DataLoader

This will load the Ministral-8B-Instruct-2410 model and prepare the training pipeline.

```python
# Initialize the dataloader
zarr_path = Path.cwd().parent.parent / "test_aifs_large.zarr"

print(f"Loading from: {zarr_path}")
print("\nNote: This will load the Ministral-8B-Instruct-2410 model (~16GB)")
print("Model will be loaded in float16 on MPS (Apple Silicon GPU) for efficiency\n")

dataloader = ClimateTextDataLoader(
    zarr_paths=str(zarr_path),
    samples_per_epoch=3,
    cache_mistral_model=True,
    device="cpu",  # Will auto-detect and use MPS if available
    seed=42
)

print(f"\n✓ DataLoader initialized successfully")
```

## Generate Training Samples

Generate 3 samples with real climate data and Ministral responses.

**Note:** Using `max_response_length=2048` to get full-length responses instead of the default 512 tokens used during training.

```python
# Generate samples with full-length responses (2048 tokens instead of default 512)
print("Generating 3 training samples...\n")
samples = dataloader.get_sample_preview(num_samples=3, max_response_length=2048)
print(f"\n✓ Generated {len(samples)} samples")
```
